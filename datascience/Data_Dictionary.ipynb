### B. Fitur Engineering (5 kolom)
Fitur-fitur berikut dihasilkan dari kombinasi fitur asli untuk memberikan informasi tambahan bagi model.

| No | Nama Fitur | Deskripsi | Formula | Tipe Data | Range |
| --- | --- | --- | --- | --- | --- |
| 20 | `Study_Efficiency` | Efisiensi belajar — skor yang didapat per jam belajar | `Previous_Scores / (Hours_Studied + 1)` | Float | Bervariasi |
| 21 | `Engagement_Score` | Skor keterlibatan — kombinasi kehadiran dan jam belajar | `Attendance × Hours_Studied / 100` | Float | Bervariasi |
| 22 | `Sleep_Study_Ratio` | Rasio jam tidur terhadap jam belajar | `Sleep_Hours / (Hours_Studied + 1)` | Float | Bervariasi |
| 23 | `Support_Score` | Skor dukungan lingkungan belajar | `(Parental_Involvement + Access_to_Resources + Internet_Access + Family_Income) / 4` | Float | 0.0–2.0 |
| 24 | `Academic_Readiness` | Skor kesiapan akademik (weighted) | `Previous_Scores/100 × 0.4 + Motivation_Level/2 × 0.3 + Tutoring_Sessions/max × 0.3` | Float | 0.0–1.0 |

### C. Target (1 kolom)

| Nama Fitur | Deskripsi | Tipe Data | Values | Cara Pembuatan |

### Data Preprocessing yang Sudah Dilakukan
